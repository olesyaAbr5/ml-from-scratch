# Предсказание прибыли СТО по обученной модели

Загружает веса, сохранённые в `Theta.txt` в результате `main.ipynb`, и предсказывает прибыль по числу автомобилей, введённому пользователем.

```python
import numpy as np

theta = np.loadtxt('Theta.txt')

x = int(input('Введите число автомобилей (тыс.): '))


profit = np.dot([1, x], theta)

print(f"Прогнозируемая прибыль СТО: {profit:.4f}")
```
