```python
data = pd.read_csv("iris_data.csv", index_col=0)
data = data[data['class'] != 'Iris-virginica']
X = data.drop('class', axis=1).to_numpy()
y = data['class']
y = np.where(y == 'Iris-setosa', 1, 0)
print(X.shape, y.shape)

```

```python
class Perception:

    def __init__(self, eta=0.1, n_iter=50, random_state=1):
        self.eta = eta
        self.n_iter = n_iter
        self.random_state = random_state

    def fit(self, X, y):
        rgen = np.random.RandomState(self.random_state)
        self.w = rgen.normal(loc=0, scale=0.01, size=X.shape[1])
        self.b = 0.0
        self.errors_ = []
        for _ in range(self.n_iter):
            errors = 0
            for xi, target in zip(X, y):
                update = self.eta * (target - self.predict(xi))
                self.w += update * xi
                self.b += update
                errors += int(update != 0.0)
            self.errors_.append(errors)
        return self

    def predict(self, X):
        return 1 if self.net_input(X) >= 0 else 0

    def net_input(self, X):
        return np.dot(X, self.w) + self.b


ob = Perception()
ob.fit(X, y)
res = ob.predict([5.1, 3.5, 1.3, 0.23])
print("setosa" if res == 1 else "versicolor")

```
